```python
import numpy as np
from scipy.integrate import solve_ivp
import pandas as pd
import matplotlib.pyplot as plt

# Parameters
Vmax1, Km1 = 18.94*60*20, 7.094e6
Vmax2, Km2 = 227.68*60*20, 1.590e6
Vmax3, Km3 = 147.526*60*20, 1.064e6

# ODE system
def cascade_ode(t, y):
    S, I1, I2, P = y

    v1 = Vmax1 * S / (Km1 + S)
    v2 = Vmax2 * I1 / (Km2 + I1)
    v3 = Vmax3 * I2 / (Km3 + I2)

    dS_dt = -v1
    dI1_dt = v1 - v2
    dI2_dt = v2 - v3
    dP_dt = v3

    return [dS_dt, dI1_dt, dI2_dt, dP_dt]

# Initial conditions
y0 = [1e8, 0.0, 0.0, 0.0]  # [S, I1, I2, P]

# Time span
t_span = (0, 60)
t_eval = np.linspace(0, 60, 61)

# Solve ODEs
solution = solve_ivp(cascade_ode, t_span, y0, t_eval=t_eval)

# Create a DataFrame with the solution
data = {
    'Time': solution.t,
    'Substrate_S': solution.y[0],
    'Intermediate_1': solution.y[1],
    'Intermediate_2': solution.y[2],
    'Product_P': solution.y[3]
}
df = pd.DataFrame(data)

# Save raw data to CSV
df.to_csv("Cas_opt.csv", index=False)
print("Raw data saved to 'Cas_Low.csv'")

# Display first few rows
df.head()
```
